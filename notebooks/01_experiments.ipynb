# Eksperymenty z kierunkiem cen

Aktualny przykład korzysta z pakietu `stock_direction`. Najpierw zainstaluj projekt zgodnie z README. Wyniki dotyczą klasyfikacji, bez deklaracji wyniku strategii. Stare notebooki są w `legacy/`.


```python
from pathlib import Path

from stock_direction.cli import main

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists())
DATA_DIR = ROOT / "data" / "raw" / "prices"
```

Każde wywołanie tworzy nowy katalog raportów. Wszystkie modele mają wspólny cel i daty oceny. Zmień `--last-test-year` świadomie, zachowując oddzielny protokół końcowej oceny.


```python
main(
    [
        "--ticker",
        "AAPL",
        "--models",
        "lr",
        "majority",
        "always_up",
        "random",
        "--data-dir",
        str(DATA_DIR),
        "--output-dir",
        str(ROOT / "reports" / "runs" / "aapl_binary_example"),
    ]
)
```

Przykład TBM: klasy -1/0/+1, horyzont 5 sesji, bariery oceniane po cenach zamknięcia. To inne zadanie niż binary; wartości accuracy nie są bezpośrednio porównywalne.


```python
main(
    [
        "--ticker",
        "AAPL",
        "--target",
        "tbm",
        "--models",
        "lr",
        "majority",
        "always_up",
        "--data-dir",
        str(DATA_DIR),
        "--output-dir",
        str(ROOT / "reports" / "runs" / "aapl_tbm_example"),
    ]
)
```

Raporty można wczytać przez `pandas.read_csv`. Ponowne wykonanie przykładu z tą samą nazwą katalogu zgłosi błąd; wybierz nową nazwę, aby zachować poprzedni wynik. Ograniczenia i dalsze usprawnienia: `docs/AUDIT.md`.
